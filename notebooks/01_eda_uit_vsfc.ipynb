```python
import pandas as pd
import uuid

# Danh sách 3 tập dữ liệu cần xử lý độc lập
splits = ['train', 'validation', 'test']

for split in splits:
    print(f"=== XỬ LÝ TẬP: {split.upper()} ===")
    
    df = pd.read_csv(f'../data/vietnamese_students_feedback/{split}.csv')
    print(f"Số lượng mẫu ban đầu: {len(df)}")
    
    # 2. Xóa các dòng bị lỗi
    df = df.dropna(subset=['sentence', 'sentiment'])
    df['sentence'] = df['sentence'].astype(str)
    df = df[df['sentence'].str.strip() != ""]
    
    df['sentiment'] = df['sentiment'].astype(int)
    valid_labels = [0, 1, 2]
    df = df[df['sentiment'].isin(valid_labels)]
    
    # 3. Chuẩn hóa Schema
    df = df.rename(columns={'sentence': 'text_raw', 'sentiment': 'label'})
    df['id'] = [str(uuid.uuid4()) for _ in range(len(df))]
    df['text_clean'] = df['text_raw']
    df['source'] = 'student_feedback'
    df['dataset_name'] = 'UIT-VSFC'
    df['split'] = split
    df['annotation_status'] = 'official'
    
    schema_cols = ['id', 'text_raw', 'text_clean', 'label', 'source', 'dataset_name', 'split', 'annotation_status']
    df = df[schema_cols]
    
    duplicates_count = df.duplicated(subset=['text_raw']).sum()
    df = df.drop_duplicates(subset=['text_raw'])
    print(f"Số mẫu trùng lặp đã xóa: {duplicates_count}")
    print(f"Số mẫu thực tế sau khi làm sạch: {len(df)}")
    
    print("Phân bố nhãn (%):")
    print(df['label'].value_counts(normalize=True) * 100)
    
    output_path = f'../data/vietnamese_students_feedback/{split}.csv'
    df.to_csv(output_path, index=False, encoding='utf-8-sig')
```
